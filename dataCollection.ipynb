```python
from datasets import load_dataset
import os

os.makedirs("./data/papers", exist_ok=True)

dataset = load_dataset(
    "jamescalam/ai-arxiv",
    split="train",
    streaming=True
)

saved = 0
target = 300

print("Starting...")

for paper in dataset:
    title   = paper.get("title", "")
    content = paper.get("content", paper.get("text", paper.get("body", "")))
    
    if len(content.strip()) < 300:
        continue

    filename = f"./data/papers/paper_{saved+1:04d}.txt"
    with open(filename, "w", encoding="utf-8") as f:
        f.write(f"Title: {title}\n\n{content}")

    saved += 1
    if saved % 25 == 0:
        print(f"  Saved {saved}/{target}...")

    if saved >= target:
        break

print(f"\nDone! {saved} papers saved")
```

```python
import os

papers_dir = "./data/papers"
files = [f for f in os.listdir(papers_dir) if f.endswith(".txt")]

total_size = 0
too_small = []
good = 0

for f in files:
    size = os.path.getsize(os.path.join(papers_dir, f))
    total_size += size
    if size < 500:
        too_small.append(f)
    else:
        good += 1

print(f"Total papers     : {len(files)}")
print(f"Healthy papers   : {good}")
print(f"Too small (<500B): {len(too_small)}")
print(f"Total size       : {total_size / 1_000_000:.1f} MB")
print(f"Avg size         : {total_size / len(files) / 1_000:.1f} KB per paper")

# Preview one paper
sample = os.path.join(papers_dir, files[0])
with open(sample, "r", encoding="utf-8") as f:
    preview = f.read(500)
print(f"\nSample preview:\n{'-'*40}\n{preview}\n{'-'*40}")
```
